Ejercico 2: Dado el diccionario datos = {'a':10, 'b':25, 'c':7, 'd':18}
a) Crea una lista con las claves cuyo valor sea mayor a 12.
b) Crea un nuevo diccionario con esas claves y sus valores al cuadrado.


```python
datos = {'a': 10, 'b': 25, 'c': 7, 'd': 18}

claves_filtradas = [clave for clave, valor in datos.items() if valor > 12]

diccionario_cuadrado = {k: v**2 for k, v in datos.items() if v > 12}

print(f"Claves con valor > 12: {claves_filtradas}")
print(f"Diccionario transformado: {diccionario_cuadrado}")
```

Ejercicio 3:  Define una función 'resumen(lista)' que retorne un diccionario con:
longitud, suma, promedio, mínimo y máximo de la lista

```python
def resumen(lista):
    if not lista:
        return None

    longitud = len(lista)
    suma = sum(lista)
    promedio = suma / longitud
    minimo = min(lista)
    maximo = max(lista)

    return {
        "longitud": longitud,
        "suma": suma,
        "promedio": promedio,
        "minimo": minimo,
        "maximo": maximo
    }

mi_lista = [10, 20, 30, 40, 50]
resultado = resumen(mi_lista)
print(resultado)
```
