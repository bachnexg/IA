#Indexación Booleana

```python
import numpy as np
datos = np.array([[5, 12, 3, 18],
[7, 2, 15, 9],
[11, 6, 20, 1]])
```

# ── Crear máscara booleana

```python
mascara = datos > 10 # array de True/False con misma shape
print(mascara)
```

# ── Aplicar máscara (filtro)

```python
mayores = datos[mascara] # devuelve 1-D con los valores True
print(mayores)
```

# ── Combinación de condiciones

```python
rango = datos[(datos >= 5) & (datos <= 15)]
print(rango) # [ 5 12 7 15 9 11 6]
```

# ── Modificar elementos con máscara

```python
copia = datos.copy()
copia[copia < 5] = 0 # reemplaza valores menores a 5 con 0
print(copia)
```

# ── np.where — condicional vectorizado

```python
resultado = np.where(datos > 10, datos, 0) # deja >10, reemplaza resto con 0
print(resultado)
```

#Estadística y Reducciones

```python
M = np.array([[4, 7, 2, 8],
[3, 11, 5, 1],
[6, 9, 0, 4]])
```

Reducción global (todos los ejes)

```python
print(M.sum()) # 60
print(M.mean()) # 5.0
print(M.max()) # 11
print(M.min())
```

 Reducción por eje

```python
print(M.sum(axis=0)) # suma por columna: [13 27 7 13]
print(M.sum(axis=1)) # suma por fila: [21 20 19]
print(M.mean(axis=0)) # media de cada columna
print(M.max(axis=1))
```

argmin / argmax — índice del extremo

```python
print(M.argmin(axis=1)) # índice del mínimo en cada fila: [2 3 2]
print(M.argmax(axis=0))
```

std y var

```python
print(M.std()) # desviación estándar global
print(M.var(axis=0))
```

keepdims — mantener dimensiones

```python
media = M.mean(axis=0, keepdims=True) # shape (1, 4) en vez de (4,)
print(media.shape) # (1, 4)
centrado = M - media
```

cumsum — suma acumulada

```python
v = np.array([1, 2, 3, 4, 5])
print(np.cumsum(v))
```

percentile

```python
print(np.percentile(M, 25)) # percentil 25 (primer cuartil)
print(np.median(M))
```

#Concatenación y Apilamiento de Arrays


```python
import numpy as np
A = np.array([[1, 2, 3],
[4, 5, 6]])
B = np.array([[7, 8, 9],
[10, 11, 12]])
```

concatenate

```python
vertical = np.concatenate([A, B], axis=0) # apilar filas
print(vertical.shape) # (4, 3)
```

vstack y hstack

```python
v1 = np.array([1, 2, 3])
v2 = np.array([4, 5, 6])
print(np.vstack([v1, v2])) # [[1 2 3], [4 5 6]] → shape (2,3)
print(np.hstack([v1, v2]))
```

stack — crea nuevo eje

```python
capas = np.stack([A, B], axis=0) # shape (2, 2, 3) — nuevo eje al frente
print(capas.shape)
```

split

```python
M = np.arange(1, 25).reshape(4, 6)
partes = np.split(M, 2, axis=1) # divide en 2 a lo largo de columnas
print(partes[0].shape) # (4, 3)
print(partes[1].shape) # (4, 3)
filas = np.vsplit(M, 4) # divide en 4 filas individuales
print(filas[0])
```

#Ordenamiento y Búsqueda

```python
v = np.array([5, 2, 8, 1, 9, 3, 7, 4, 6])
```

sort

 argsort

```python
v = np.array([5, 2, 8, 1, 9, 3])
idx = np.argsort(v) # [3 1 5 0 2 4] — índices que ordenarían v
print(v[idx])
```

Reordenar dos arrays en paralelo


```python
nombres = np.array(['Carlos','Ana','Diana','Pedro','Eva','Luis'])
notas = np.array([75, 92, 88, 61, 95, 78])
orden = np.argsort(notas)[::-1] # índices de mayor a menor
print(nombres[orden]) # ['Eva' 'Ana' 'Diana' 'Luis' 'Carlos' 'Pedro']
print(notas[orden])
```
